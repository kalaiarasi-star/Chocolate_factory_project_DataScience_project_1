```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
#csv file
df = pd.read_csv("chocolate_factory.csv")
# Convert Date column
df["Date"] = pd.to_datetime(df["Date"])

print("\n----------- CHOCOLATE FACTORY DATA ----------")
print(df)

print("\n---------- FIRST 5 RECORDS ----------")
print(df.head())

print("\n---------- DATASET SHAPE ----------")
print(df.shape)

print("\n---------- COLUMN NAMES ----------")
print(df.columns.tolist())

print("\n---------- DATA TYPES ----------")
print(df.dtypes)

print("\n---------- MISSING VALUES ----------")
print(df.isnull().sum())

print("\n---------- STATISTICAL SUMMARY ----------")
print(df.describe())

total_production = df["Production_Units"].sum()

print("\nTotal Production Units:", total_production)

total_sales = df["Sales_Units"].sum()

print("Total Sales Units:", total_sales)

total_revenue = df["Revenue"].sum()

print("Total Revenue:", total_revenue)

total_defects = df["Defective_Units"].sum()

print("Total Defective Units:", total_defects)

production = df.groupby("Product")["Production_Units"].sum()

print("\n---------- PRODUCT-WISE PRODUCTION ----------")
print(production)

highest_production = production.idxmax()

print("\nHighest Production Product:")
print(highest_production)

sales = df.groupby("Product")["Sales_Units"].sum()

print("\n---------- PRODUCT-WISE SALES ---------")
print(sales)

best_selling = sales.idxmax()

print("\nBest Selling Product:")
print(best_selling)

revenue = df.groupby("Product")["Revenue"].sum()

print("\n---------- PRODUCT-WISE REVENUE ----------")
print(revenue)

highest_revenue = revenue.idxmax()

print("\nHighest Revenue Product:")
print(highest_revenue)

defects = df.groupby("Product")["Defective_Units"].sum()

print("\n---------- PRODUCT-WISE DEFECTS ----------")
print(defects)

highest_defects = defects.idxmax()

print("\nProduct with Highest Defective Units:")
print(highest_defects)

quality = df.groupby("Product")["Quality_Score"].mean()

print("\n---------- AVERAGE QUALITY SCORE ----------")
print(quality.round(2))

best_quality = quality.idxmax()

print("\nHighest Quality Product:")
print(best_quality)

sales_percentage = df.groupby("Product")["Sales_Percentage"].mean()

print("\n---------- SALES PERCENTAGE ----------")
print(sales_percentage.round(2))

highest_sales_percentage = sales_percentage.idxmax()

print("\nHighest Sales Percentage:")
print(highest_sales_percentage)

defect_rate = df.groupby("Product")["Defect_Rate"].mean()

print("\n---------- DEFECT RATE ----------")
print(defect_rate.round(2))

lowest_defect_rate = defect_rate.idxmin()

print("\nLowest Defect Rate:")
print(lowest_defect_rate)

df["Month"] = df["Date"].dt.strftime("%b")

monthly_production = df.groupby("Month")["Production_Units"].sum()

print("\n----------- MONTHLY PRODUCTION ----------")
print(monthly_production)

monthly_sales = df.groupby("Month")["Sales_Units"].sum()

print("\n---------- MONTHLY SALES ----------")
print(monthly_sales)

monthly_revenue = df.groupby("Month")["Revenue"].sum()

print("\n---------- MONTHLY REVENUE ----------")
print(monthly_revenue)

sns.set_theme(style="whitegrid")

plt.figure(figsize=(10, 6))

sns.barplot(
    x=production.index,
    y=production.values
)

plt.title("Total Production by Chocolate Product")
plt.xlabel("Product")
plt.ylabel("Production Units")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))

sns.barplot(
    x=sales.index,
    y=sales.values
)

plt.title("Total Sales by Chocolate Product")
plt.xlabel("Product")
plt.ylabel("Sales Units")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))

sns.barplot(
    x=revenue.index,
    y=revenue.values
)

plt.title("Total Revenue by Chocolate Product")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))

sns.barplot(
    x=defects.index,
    y=defects.values
)

plt.title("Defective Units by Product")
plt.xlabel("Product")
plt.ylabel("Defective Units")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))

sns.barplot(
    x=quality.index,
    y=quality.values
)

plt.title("Average Quality Score by Product")
plt.xlabel("Product")
plt.ylabel("Quality Score")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))

sns.barplot(
    x=sales_percentage.index,
    y=sales_percentage.values
)

plt.title("Average Sales Percentage by Product")
plt.xlabel("Product")
plt.ylabel("Sales Percentage (%)")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))

sns.barplot(
    x=defect_rate.index,
    y=defect_rate.values
)

plt.title("Average Defect Rate by Product")
plt.xlabel("Product")
plt.ylabel("Defect Rate (%)")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()

month_order = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

monthly_production = monthly_production.reindex(month_order)

plt.figure(figsize=(10, 6))

plt.plot(
    monthly_production.index,
    monthly_production.values,
    marker="o"
)

plt.title("Monthly Chocolate Production")
plt.xlabel("Month")
plt.ylabel("Production Units")

plt.tight_layout()
plt.show()

monthly_sales = monthly_sales.reindex(month_order)

plt.figure(figsize=(10, 6))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o"
)

plt.title("Monthly Chocolate Sales")
plt.xlabel("Month")
plt.ylabel("Sales Units")

plt.tight_layout()
plt.show()

monthly_revenue = monthly_revenue.reindex(month_order)

plt.figure(figsize=(10, 6))

plt.plot(
    monthly_revenue.index,
    monthly_revenue.values,
    marker="o"
)

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df,
    x="Production_Units",
    y="Sales_Units",
    hue="Product",
    s=100
)

plt.title("Production vs Sales")
plt.xlabel("Production Units")
plt.ylabel("Sales Units")

plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df,
    x="Price_Per_Unit",
    y="Sales_Units",
    hue="Product",
    s=100
)

plt.title("Price vs Sales")
plt.xlabel("Price per Unit")
plt.ylabel("Sales Units")

plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df,
    x="Production_Units",
    y="Defective_Units",
    hue="Product",
    s=100
)

plt.title("Production vs Defective Units")
plt.xlabel("Production Units")
plt.ylabel("Defective Units")

plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 6))

sns.histplot(
    df["Revenue"],
    bins=10,
    kde=True
)

plt.title("Revenue Distribution")
plt.xlabel("Revenue")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()

numeric_columns = [
    "Production_Units",
    "Sales_Units",
    "Price_Per_Unit",
    "Revenue",
    "Defective_Units",
    "Employee_Count",
    "Quality_Score",
    "Unsold_Units",
    "Sales_Percentage",
    "Defect_Rate"
]

correlation = df[numeric_columns].corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Chocolate Factory Correlation Heatmap")

plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="Product",
    y="Revenue"
)

plt.title("Revenue Distribution by Product")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()

print("\n")
print("==============================================")
print("       CHOCOLATE FACTORY EDA ANSWERS")
print("==============================================")

print("\n1. Which product has the highest production?")
print("   Answer:", highest_production)

print("\n2. Which product has the highest sales?")
print("   Answer:", best_selling)

print("\n3. Which product generates the highest revenue?")
print("   Answer:", highest_revenue)

print("\n4. Which product has the highest defective units?")
print("   Answer:", highest_defects)

print("\n5. Which product has the highest quality score?")
print("   Answer:", best_quality)

print("\n6. Which product has the highest sales percentage?")
print("   Answer:", highest_sales_percentage)

print("\n7. Which product has the lowest defect rate?")
print("   Answer:", lowest_defect_rate)

print("\n8. Total production:")
print("   Answer:", total_production, "units")

print("\n9. Total sales:")
print("   Answer:", total_sales, "units")

print("\n10. Total revenue:")
print("    Answer:", total_revenue)

print("\n11. Total defective units:")
print("    Answer:", total_defects)

print("\n==============================================")
print("              EDA COMPLETED")
print("==============================================")
```
